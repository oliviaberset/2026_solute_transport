## The data

The NAWA-Fracht dataset includes DOC measurements (variable `'DOC'`). As a starting point, you can focus on the stations:

- Alpthal, Luempenenbach (LU): very small experimental forested catchment at 1100 masl
- Sellenbodenbach, Neuenkirch (NE): small agricultural catchment at 515 masl
- Glatt, Rheinsfelden (GL): medium-size catchment at 340 masl with large agricultural and urban areas

because they have comparable DOC values but different elevations and land uses. You are welcome to explore DOC data at any other station too.

![NAWA-Fracht stations map](nawa-fracht_stations_map.jpg)

Data files provided for the exercise

- **NAWA-Fracht_data_14d.csv**: concentrations in a two-weekly rhythm (composite samples) with status and comments concerning reliability of sample and  chemical analysis

- **NAWA-Fracht_basin_properties.csv**: csv - file containing the names and coordinates of monitoring stations, mean river discharge, river length, basin area,  average altitude and land use.
